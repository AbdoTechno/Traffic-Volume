# 1. Problem Statement and Project Scoping

## Metro Interstate Traffic Volume Prediction & Analysis

---

### Project Summary
Urban traffic congestion is a major challenge for modern metropolitan infrastructure. Interstate 94 (I-94) is a critical transportation corridor connecting Minneapolis and St. Paul in Minnesota. Traffic flow on this interstate exhibits complex non-linear patterns governed by time of day, day of week, seasonal shifts, national holidays, and severe Midwestern weather conditions (such as heavy snowfall, rainstorms, and sub-zero temperatures).

This project aims to formulate a data-driven machine learning framework to analyze historical traffic patterns and build accurate regression models capable of predicting hourly westbound traffic volume.

---

## 1. Business Problem & Motivation

### The Challenge
Unanticipated traffic congestion leads to substantial economic losses, increased carbon emissions, heightened accident risks, and severe transit delays for both daily commuters and commercial freight transport.

### Key Stakeholders and Impact
1. **Departments of Transportation (DOT):** Need reliable forecasts to optimize lane utilization, dynamically schedule highway maintenance, and manage emergency response routing.
2. **Commuters & Logistics Operators:** Rely on accurate traffic volume estimates to select optimal travel windows and minimize transit times.
3. **Urban & Environmental Planners:** Utilize traffic pattern insights to assess environmental impact and plan long-term highway infrastructure.

---

## 2. Research Questions

To solve this problem effectively, this project investigates the following primary research questions:

1. **Temporal Periodicity:** What are the exact rush-hour dynamics across weekdays versus weekends, and how do regional/federal holidays suppress or shift peak traffic volume?
2. **Weather Sensitivity:** How significantly do adverse meteorological events (e.g., rainfall, snowfall, freezing temperatures, cloud cover) degrade traffic capacity and throughput?
3. **Predictive Performance:** Can machine learning regression models (ranging from regularized linear baselines to ensemble gradient boosting algorithms like LightGBM and XGBoost) predict hourly traffic volume with high precision ($R^2 > 0.90$)?

---

## 3. Project Objectives & Success Criteria

### Primary Objectives
- **Data Ingestion & Hygiene:** Build an automated pipeline to extract the raw dataset from the UCI Machine Learning Repository, detect anomalous sensor records (such as 0 Kelvin temperature spikes), and prepare clean temporal features.
- **Exploratory Data Analysis (EDA):** Perform deep visual and statistical analysis to uncover correlations between external environmental variables and traffic flow.
- **Feature Engineering:** Design domain-specific features, including cyclical trigonometric time encodings (Sine/Cosine of hour/month) and weather category encodings.
- **Model Development & Benchmarking:** Train and compare multiple regression models using time-based validation splits to prevent temporal look-ahead leakage.
- **Explainability & Recommendations:** Quantify feature importance (e.g., via SHAP or tree importance) to deliver actionable insights for traffic management.

### Quantitative Success Metrics
The models will be evaluated using standard continuous regression metrics:
- **Coefficient of Determination ($R^2$ Score):** Target $\ge 0.90$
- **Mean Absolute Error (MAE):** Quantifies average vehicle forecast deviation in units of vehicles/hour.
- **Root Mean Squared Error (RMSE):** Penalizes larger outlier prediction errors.

## 4. Dataset Overview & Scope

The data source for this project is the **Metro Interstate Traffic Volume** dataset, provided by the Minnesota Department of Transportation (MN DoT) ATR station 301 and hosted on the UCI Machine Learning Repository.

| Characteristic | Specification |
| :--- | :--- |
| **Repository** | UCI Machine Learning Repository (ID: 492) |
| **Target Variable** | `traffic_volume` (Hourly Westbound Vehicle Count) |
| **Total Instances** | 48,204 hourly records |
| **Number of Features** | 8 input features (Weather, Calendar, and Datetime) |
| **Time Horizon** | 2012 to 2018 (Hourly granularity in local CST) |
| **Missing Values** | 0 missing values reported out-of-the-box |

### Feature Definitions
- `holiday`: Categorical indicator for US national holidays and Minnesota regional events.
- `temp`: Average hourly ambient temperature in Kelvin ($K$).
- `rain_1h`: Hourly rainfall accumulation in millimeters ($mm$).
- `snow_1h`: Hourly snowfall accumulation in millimeters ($mm$).
- `clouds_all`: Percentage of cloud coverage ($0-100\%$).
- `weather_main`: High-level weather category (e.g., Clear, Clouds, Rain, Snow).
- `weather_description`: Granular descriptive condition (e.g., light rain, heavy snow).
- `date_time`: Hourly timestamp formatted as `YYYY-MM-DD HH:MM:SS` (CST).
- `traffic_volume`: Westbound hourly volume (Target variable).
